**Q1. reshape() vs resize()**

              reshape()	                      resize()

1. Changes the shape of an array	    1. Changes the size & shape of an array
2. Does not change the original 	    2. Changes the original array when       array                                    used as arr.resize()
3. Total number of elements must 	    3. Number of elements can increase or
  remain the same                         decrease
4. Returns a new reshaped array	      4. Can modify the original array directly


* **Examples of reshape and resize**



```python
import numpy as np
```

```python
# reshape()

arr = np.arange(6)   # 1D array
#new_arr = arr.reshape(2, 3)   # 2D array

print("Reshape array 2D - ", arr.reshape(2, 3))  # after reshaping it only change temporary
print("Original array 1D - ", arr)
```

```python
# resize()

b = np.arange(6)
print("Original array 1D - ", b)

b.resize(2, 3)
print("Resize array 2D - ", b)
# same array changes permanantly
```

1. Add a number in 1D array  (add a number to every element of a 1D array)

```python
a = np.arange(5)          # a = np.arange(1,5)  start and stop both values for 1-D
print("Original array - ", a)

addition = a + 5
print("O/P -", addition)
```

```python
b = np.arange(5)

add_element = np.insert(b, 5, 10)

print(add_element)
```

2. Add a row in 2D array

```python
c = np.array([
    [1, 2, 3],
    [4, 5, 6]])

row = [7, 8, 9]
new_arr = np.vstack((c, row))

print(new_arr)
```

3. Add a column in 2D array

```python
d = np.array([
    [1, 2, 3],
    [4, 5, 6]])

col = np.array([
    [7],
    [8] ])

e = np.hstack((d, col))   # adding column
print(e)
```

4. Remove a number from 1D array

```python
f = np.arange(6)
print("Original array - ", f)

f = np.delete(f, 3)       # delete value at index=3
print("O/P - ", f)
```

```python
g = np.array([10, 20, 30, 40, 50])
print("Original array - ", g)

g = g[g != 40]
print("O/P - ", g)
```

5. Remove a row from 2D array

```python
h  = np.array([
    [1, 2, 3],
    [4, 5, 6]])
print("Original array - ", h)
h = np.delete(h, 1, axis=0)    # 1 = row index (row index starts with 0,1,2...) axis = 0 for row
print("O/P -", h)
```

6. Remove a column from 2D array

```python
i  = np.array([
    [1, 2, 3],
    [4, 5, 6]])
print("Original array - ", i)
i = np.delete(i, 1, axis=1)    # 1 = column index (col index starts with 0,1,2...) axis = 1 for column
print("O/P -", i)
```

7. Concatenate 2 arrays
