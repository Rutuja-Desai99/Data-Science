**Q1. What is difference between list and numpy array?**

	             	                      

* **Feature             Python List                      NumPy Array**

1. Library	      # Built into Python	              # Requires numpy library

2. Data types	    # Can contain different types	    # Usually same data type

3. Speed	        # Slower for numerical operations	# Faster for numerical ops

4. Memory	        # Uses more memory	              # More memory efficient

5. Maths ops   	  # Need loops	                    # Direct element-wise ops

6. Dimensions	    # Mainly 1D, nested lists for     # multi-dimensional arrays
                    multiple dimensions

7. Best for	      # General-purpose data	          # Numericalscientific
                                                       calculations



* **Example Python list and Numpy array**



```python
# Python List

lst = [1, 2, 3, 4]
result = [x * 2 for x in lst]
print(result)
```

```python
# With a list
[1, 2, 3] * 2
```

```python
# NumPy Array

import numpy as np
arr = np.array([1, 2, 3, 4])
result = arr * 2
print(result)
```

```python
# with NumPy

np.array([1, 2, 3]) * 2
```

1. Create 4x4 array of all True values

```python
a = np.full((4,4),'True')   # using shape and fill value
a
```

```python
z = np.full((4,4),'A')   # using shape and fill value can change value
z
```

2. Create 4x4 array of all False values

```python
c = np.full((4,4),'False')   # using shape and fill value
c
```

```python
# using zeros as false
d = np.zeros((4, 4), dtype=bool)
d
```

3. Create 4x4 array of Digonal elements as 1's and Non diagonal elements as 0's

```python
# Using eye
e = np.eye(4,dtype=int)          # 4 consider as row and column as we need 4x4 matrix
                                 # and index = default 0

#e = np.eye(4,4,0,dtype=int)     # row,column,digonal index = default 0
e
```

4. Create 4x4 array of Digonal elements as True and Non diagonal elements as False
