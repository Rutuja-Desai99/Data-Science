```python
import numpy as np
```

1.Create 4x4 array of all True values

```python
a = np.full((4,4),'True')   # using shape and fill value
a
```

```python
# using ones as true
b = np.ones((4, 4), dtype=bool)
b
```

2.Create 4x4 array of all False values

```python
c = np.full((4,4),'False')   # using shape and fill value
c
```

```python
# using zeros as false
d = np.zeros((4, 4), dtype=bool)
d
```

3.Create 4x4 array of Digonal elements as 1's and Non diagonal elements as 0's

```python
e = np.eye(4,dtype=int)          # 4 consider as row and column as not mentioned different values
                                 # and index = default 0

#e = np.eye(4,4,0,dtype=int)     # row,column,digonal index = default 0
e
```

4.Create 4x4 array of Digonal elements as True and Non diagonal elements as False

```python
f = np.eye(4,dtype=bool)          # 4 consider as row and column and index =default 0
#f = np.eye(4,4,0,dtype=bool)      # row,column,digonal index =default 0
f
```

5.Create 4x4 array of Digonal elements as False and Non diagonal elements as True

```python
f = ~np.eye(4,dtype=bool)      # row,column,digonal index =default 0
#f = ~np.eye(4,4,0,dtype=bool)      # row,column,digonal index =default 0
f
```
